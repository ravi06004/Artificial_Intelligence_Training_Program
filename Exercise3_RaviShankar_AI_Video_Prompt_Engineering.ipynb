Exercise 3: Generative AI Prompt Engineering for Video Synthesis

Selected AI Video-Generation Tool
        Tool Name: Openart ai (Image-to-Video)
Images Used
   Three images created in Exercise 2:
   Image 1 – First source image
   Image 2 – Second source image
   Image 3 – Blended image

Exact Video Generation Prompt
"Create a visually appealing 3-second cinematic video using the three provided reference images. Begin with the first source image displayed clearly in the frame. Gradually introduce the second source image through a smooth visual transition, blending both images naturally to create the final blended composition. Maintain the original subjects, colors, textures, and important visual details of all three images. Use subtle motion and smooth transitions to make the transformation look realistic and seamless. End with the blended image fully visible and centered in the frame. Ensure high visual quality, consistent lighting, and no unwanted distortions."

Camera and Transition Instructions
0–1 second: Display the first source image with a slow camera zoom-in.
1–2 seconds: Smooth cross-dissolve transition from the first image to the second image, gradually introducing the blended composition.
2–3 seconds: Reveal the final blended image with a subtle zoom-out and hold the final frame.
Camera Movement: Slow zoom-in followed by a gentle zoom-out.

Transition Style: Smooth cross-dissolve with seamless image blending.

Visual Style: Cinematic, realistic, smooth, and high-resolution.

Expected Output: A 3-second video showing a smooth transformation of the two source images into the final blended image.
