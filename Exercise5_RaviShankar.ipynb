```python
import gradio as gr
```

```python
def chatbot(message, history):
    question = message.lower().strip()

    if "hello" in question or "hi" in question:
        return "Hello! Welcome to my AI chatbot. How can I help you?"

    elif "artificial intelligence" in question:
        return (
            "Artificial Intelligence (AI) enables computers "
            "to perform tasks that normally require human intelligence."
        )

    elif "machine learning" in question:
        return (
            "Machine Learning is a branch of AI that allows "
            "computers to learn from data and make predictions."
        )

    elif "gradio" in question:
        return (
            "Gradio is a Python library used to create "
            "interactive web interfaces for AI applications."
        )

    else:
        return (
            "Thank you for your question! "
            "Please ask me about AI, Machine Learning, or Gradio."
        )
```

```python
demo = gr.ChatInterface(
    fn=chatbot,
    title="🤖 My AI Chatbot",
    description=(
        "Welcome to my AI chatbot! "
        "Ask questions and get instant conversational responses."
    ),
    examples=[
        "What is Artificial Intelligence?",
        "Explain Machine Learning.",
        "What is Gradio?"
    ]
)
```

```python
demo.launch(share=True)
```
