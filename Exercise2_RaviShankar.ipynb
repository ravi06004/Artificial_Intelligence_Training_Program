```python
# Import OpenCV for image processing and Matplotlib for displaying images.
import cv2
print(cv2.__version__)
import matplotlib.pyplot as plt
```

```python
# Load the two input images
img1 = cv2.imread('Exercise2_RaviShankar_Original_Image1.jpg')
img2 = cv2.imread('Exercise2_RaviShankar_Original_Image2.jpg')

# Check whether images were loaded successfully
if img1 is None or img2 is None:
    raise FileNotFoundError("Please check the image file paths.")

# Save copies of the original input images
cv2.imwrite('Original_Image1.png', img1)
cv2.imwrite('Original_Image2.png', img2)

print("Both images loaded successfully.")

# Display original image dimensions
print("Image 1 dimensions:", img1.shape)
print("Image 2 dimensions:", img2.shape)
```

```python
# Get the dimensions of the first image
height, width = img1.shape[:2]

# Resize the second image to match the first image
img2_resized = cv2.resize(img2, (width, height))

# Verify the dimensions after resizing
print("Image 1 resized dimensions:", img1.shape)
print("Image 2 resized dimensions:", img2_resized.shape)
```

```python
# Perform weighted blending of the two images
blended_image = cv2.addWeighted(
    img1, 0.5,
    img2_resized, 0.5,
    0
)

# Save the blended output image
cv2.imwrite('Blended_Image.png', blended_image)

print("Blended image successfully created and saved.")
```

```python

# Convert BGR images to RGB for correct display
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended_image, cv2.COLOR_BGR2RGB)

# Create a figure with three subplots
plt.figure(figsize=(15, 5))

# Display the first original image
plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title('Original Image 1')
plt.axis('off')

# Display the second original image
plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title('Original Image 2')
plt.axis('off')

# Display the blended image
plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title('Blended Image')
plt.axis('off')

# Adjust layout and display all images
plt.tight_layout()
plt.show()
```
