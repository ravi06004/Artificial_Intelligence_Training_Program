```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate random numerical data
np.random.seed(42)

data = {
    'f1': np.random.randint(1, 100, 10),
    'f2': np.random.randint(1, 100, 10),
    'f3': np.random.randint(1, 100, 10),
    'f4': np.random.randint(1, 100, 10)
}

# Create DataFrame
df = pd.DataFrame(data)

# Step 2: Generate scatter plot (f2 vs f3)
plt.figure(figsize=(8, 5))
plt.scatter(df['f2'], df['f3'], color='blue', marker='o')

plt.xlabel('f2')
plt.ylabel('f3')
plt.title('Scatter Plot: f2 vs f3')
plt.grid(True)

plt.show()

# Step 3: Calculate row-wise arithmetic mean
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display complete DataFrame
print("Final DataFrame:")
print(df)

# Step 4: Export DataFrame to Excel
df.to_excel('Exercise1_RaviShankar.xlsx', index=False)

print("\nData successfully exported to Exercise1_RaviShankar.xlsx")
```
